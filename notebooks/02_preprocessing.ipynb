## 4.Preprocessing

The preprocessing functions and classes are implemented in the `src/preprocessing.py` module.

In this notebook, the `preprocess_data()` function is imported from `preprocessing.py` and used to apply the preprocessing pipeline to the Titanic datasets.

The preprocessing pipeline consists of three main classes:

* **`AgeImputer`** — handles missing values in the `Age` column using the mean value.
* **`FeatureEncoder`** — applies One-Hot Encoding to the `Embarked` and `Sex` categorical features.
* **`FeaturesDropper`** — removes unnecessary features such as `Name`, `Ticket`, `Fare`, `Cabin`, `Sex`, and `Embarked`.

These preprocessing steps are combined into a single `Pipeline` inside the `preprocess_data()` function.

The processed training and test datasets are then saved using the `save_processed_train()` and `save_processed_test()` functions.

```text
raw train/test data
        ↓
   preprocess_data()
        ↓
    AgeImputer
        ↓
  FeatureEncoder
        ↓
 FeaturesDropper
        ↓
processed train/test data
```




````python
from src.preprocessing import preprocess_data,save_processed_train, save_processed_test
clean_df_train=preprocess_data(df_train)
````
